### Exercícios

Calcule o índice de massa corporal, entrando com os valores de massa do indíviduo e estatura.

```python
# Exercício  - Calcule IMC
massa = float(input("Entre com o peso em quilogramas: "))
altura = float(input("Entre com a estatura em metros: "))
imc = (massa) / (altura**2)

print(f"O seu Índice de Massa Corporal (IMC) é {imc:.2f}")
```

# Condições

## if

As condições servem para selecionar quando uma parte do programa deve ser ativada e quando deve ser simplesmente ignorada.

```python

,# Exemplo de uso de if
a = int(input("Entre com um número: "))
b = int(input("Entre com um segundo número: "))
if a > b:
    print("O primeiro valor é maior.")
if b > a:
    print("O segundo valor é maior.")
```

```python
# Exemplo de uso de if com else
a = int(input("Entre com um número: "))
b = int(input("Entre com um segundo número: "))
if a > b:
    print("O primeiro valor é maior.")
else:
    print("O segundo valor é maior.")
```

```python
# Exemplo de par ou ímpar
numero = int(input("Entre com um valor inteiro: "))

if numero % 2 == 0:
    print("par")
else:
    print("ímpar.")
```

```python
# Exercício  - Calcule IMC com if
massa = float(input("Entre com o peso em quilogramas: "))
altura = float(input("Entre com a estatura em metros: "))
imc = (massa) / (altura**2)

print(f"O seu Índice de Massa Corporal (IMC) é {imc:.2f}")

# Adicionando a interpretação do valor do IMC
if imc < 18.5:
    print("Classificação: Abaixo do peso")
elif 18.5 <= imc < 25:
    print("Classificação: Peso normal")
elif 25 <= imc < 30:
    print("Classificação: Sobrepeso")
elif 30 <= imc < 35:
    print("Classificação: Obesidade Grau I")
elif 35 <= imc < 40:
    print("Classificação: Obesidade Grau II")
else:
    print("Classificação: Obesidade Grau III (Mórbida)")
```

## Repetições

Repetições representam a base de vários programas. São utilizadas para executar a parte de um programa várias vezes. Por exemplo imprimir na tela a sequencia de números de 1 a 15.

```python
x = 1
while x <= 15:
    print(x)
    x = x + 1
```

```python
#Exemplo para ilustrar
contador = 0

while contador < 4:
    print("Dentro do while")
    contador = contador + 1
else:
    print("Dentro do else")
```

### Acumuladores

```python
# Este exemplo conta quantos números pares tem dentro de um intervalo, o contador é um acumulador e o num1 é um contador
num1 = int(input('Entre com o valor inicial: '))
num2 = int(input('Entre com o valor final: '))
contador_pares = 0
while num1 <= num2:
    if num1 % 2 == 0:
        contador_pares = contador_pares + 1
    num1 = num1 + 1
print(contador_pares)
```

```python
# Calcula o fatorial de um número inteiro positivo, resultado é um acumulador e n é o contador

n = int(input("Entre com um valor inteiro: "))
num = n
resultado = 1

if n < 0:
    print("Não é possível calcular fatorial de números negativos")
elif n == 0:
    print("O fatorial de 0 é 1")
else:
    while n > 1:
        resultado = resultado * n
        n = n -1
    print("O valor do fatorial de", num, "é", resultado)
```

### Interrompendo a repetição

```python
# Calculadora de somar, inserindo o conceito de break
soma = 0
while True:
    valor = int(input("Digite um número para somar ou 0 para sair: "))
    if valor == 0:
        break
    soma = soma + valor
print(soma)
```

#### Exercício

Escreva um programa que leia números inteiros do teclado. O programa deve ler os números até que o usuário digite 0 (zero). No final da execução, exiba a quantidade de número digitados, assim como a soma e a média aritmética.

```python
# Calculadora de somar e média
soma = 0
n = 0
while True:
    valor = int(input("Digite um número para somar ou 0 para sair: "))
    if valor == 0:
        break
    soma = soma + valor
    n = n + 1
print("A soma dos números digitados foi " + str(soma))
print("O número de valores digitados foi " + str(n))
print("A média aritmética dos números digitados foi "+ str(soma/n))
```

```python
# Usando o while
cor = ["vermelho", "verde", "azul", "amarelo"]
i = 0
while i < len(cor):
    print(cor[i])
    i += 1
```

```python
# Usando for e range
cor = ["vermelho", "verde", "azul", "amarelo"]
for i in range(len(cor)):
    print(cor[i])
```

## Desafio - Crescimento Microbiano


```python
# Desafio: Estimativa de Crescimento Exponencial
print("--- Simulador de Crescimento Celular ---")

# 1. Entrada de dados
inculo_inicial = int(input("Digite a quantidade inicial de células (inóculo): "))
populacao_alvo = int(input("Digite a população alvo desejada: "))

# 2. Inicialização das variáveis
populacao_atual = inculo_inicial
geracoes = 0

# 3. Laço de repetição: continua multiplicando enquanto não atingir o alvo
while populacao_atual < populacao_alvo:
    populacao_atual = populacao_atual * 2  # Acumulador multiplicativo (a população dobra)
    geracoes = geracoes + 1                # Contador de ciclos (gerações)

# 4. Saída de resultados
print("\n--- Resultados da Simulação ---")
print(f"Gerações necessárias: {geracoes}")
print(f"População final alcançada: {populacao_atual} células")
```
